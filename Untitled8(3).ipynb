```python
import pandas as pd

path = "/content/RideSafeCycleworks_GL_FY2025(1).xlsx"

xls = pd.ExcelFile(path)
print(xls.sheet_names)

gl = pd.read_excel(path, sheet_name="GL_Detail", dtype={"GL_Account_Number": str})

print(gl.shape)
gl.head()
```

```python
print(gl["Amount_Credit_Debit_Indicator"].unique())

gl["AmountSigned"] = gl.apply(
    lambda row: row["Amount"] if str(row["Amount_Credit_Debit_Indicator"]).strip().lower().startswith("d")
    else -row["Amount"],
    axis=1)
```

```python
keys = ["GL_Account_Number", "Business_Unit_Code", "Period"]

gl_activity = (
    gl
    .groupby(keys, as_index=False)
    .agg(Activity=("AmountSigned", "sum"), Line_Count=("AmountSigned", "size"))
)
```

```python
tb = pd.read_excel(path, sheet_name="Trial_Balance", dtype={"GL_Account_Number": str})
rf = tb.merge(
    gl_activity,
    on=keys,
    how="left",
    indicator=True,
    validate="one_to_one",
)
print(rf["_merge"].value_counts())
```

```python
rf[["Activity", "Line_Count"]] = rf[["Activity", "Line_Count"]].fillna(0)

rf["Diff"] = (rf["Amount_Ending"] - rf["Amount_Beginning"] - rf["Activity"]).round(2)
print("# of rows that don't tie:", (rf["Diff"] != 0).sum())
```

```python
print(rf.shape)
rf[['GL_Account_Number', 'Business_Unit_Code', 'Period', 'Diff']].head()
```

```python
pivot_table = rf.pivot_table(index='GL_Account_Number', columns='Period', values='Diff', aggfunc='sum')
pivot_table.head()
```

```python
# TB rows with no GL match: already in rf, where Line_Count is 0
# (the same rows as rf["_merge"] == "left_only")
tb_no_activity = rf[rf["Line_Count"] == 0]
```

```python
gl_check = gl_activity.merge(tb, on=keys, how="left", indicator=True)
gl_not_in_tb = gl_check[gl_check["_merge"] == "left_only"]
```

```python
# Accounts with a balance but no GL lines all year
lines = rf.groupby("GL_Account_Number")["Line_Count"].sum()
print(lines[lines == 0])
```

```python
accounts_per_period = tb_no_activity.groupby('Period')['GL_Account_Number'].nunique()
period_with_highest_no_activity = accounts_per_period.idxmax()
max_accounts = accounts_per_period.max()

print(f"The period with the highest number of accounts with no GL activity is '{period_with_highest_no_activity}' with {max_accounts} accounts.")
```

```python
print(f"Number of GL accounts with activity that had no matching account/business unit/period line in the Trial Balance: {len(gl_not_in_tb)}")
if not gl_not_in_tb.empty:
    print("GL Account Numbers with activity and no matching TB line:")
    print(gl_not_in_tb['GL_Account_Number'].unique())
```

```python
accounts_with_no_gl_lines = lines[lines == 0].index.tolist()
accounts_with_no_gl_lines.sort()
print("Four accounts that carry a balance all year but have no GL lines at all (in ascending order):")
for acc in accounts_with_no_gl_lines:
    print(f"* {acc}")
```
